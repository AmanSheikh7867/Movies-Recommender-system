```python
import ast
def convert(obj):
    L = []
    for i in ast.literal_eval(obj):
        L.append(i['name'])
    return L
    
```

```python
movies['genres'] = movies['genres'].apply(convert)
```

```python
movies.head()
```

```python
movies['keywords'] = movies['keywords'].apply(convert)
```

```python
 def convert3(obj):
    L = []
    counter = 0
    for i in ast.literal_eval(obj):
        if counter != 3:
            L.append(i['name'])
            counter +=1
        else:
            break
    return L
    
```

```python
movies['cast'] = movies['cast'].apply(convert3)
```

```python
movies.head()
```

```python
def fetch_director(obj):
    L = []
    for i in ast.literal_eval(obj):
        if i['job'] == 'Director' :
            L.append(i['name'])
            break
    return L
```

```python
movies['crew'] = movies['crew'].apply(fetch_director)
```

```python
# converting string of overview into the list
movies['overview'] = movies['overview'].apply(lambda x:x.split())
```

```python
# remove the space between the words

movies['genres'] = movies['genres'].apply(lambda x:[i.replace(" "," ") for i in x])
movies['keywords'] = movies['keywords'].apply(lambda x:[i.replace(" "," ") for i in x])
movies['cast'] = movies['cast'].apply(lambda x:[i.replace(" "," ") for i in x])
movies['crew'] = movies['crew'].apply(lambda x:[i.replace(" "," ") for i in x])
```

```python
# concatinate all columns in to one , tags 

movies['tags'] = movies['overview'] + movies['genres'] + movies['keywords'] + movies['cast'] + movies['crew']
```

```python
new_df['tags'] = new_df['tags'].apply(lambda x:" ".join(x))
```

```python
# Import NLTK, a Natural Language Processing (NLP) library.
# We use NLTK here to process and normalize the words in our movie tags.

import nltk
```

```python
# Import PorterStemmer.
# PorterStemmer reduces words to their root/stem form.
#
# Example:
# "playing"  → "play"
# "played"   → "play"
# "plays"    → "play"
#
# This helps us treat different forms of the same word as one feature.

from nltk.stem.porter import PorterStemmer

# Create a PorterStemmer object.
# We will use this object to apply stemming to individual words.

ps = PorterStemmer()
```

```python
def stem(text):
    # Create an empty list to store the stemmed words.
    y = []

    # Split the complete text into individual words.
    #
    # Example:
    # "love playing football"
    #        ↓
    # ["love", "playing", "football"]

    for i in text.split():

        # Apply stemming to each individual word
        # and add the stemmed word to our list.
        #
        # Example:
        # "playing" → "play"

        y.append(ps.stem(i))

    # Join all the stemmed words back together into one string.
    #
    # Example:
    # ["love", "play", "football"]
    #        ↓
    # "love play football"

    return " ".join(y)
```

```python
# Apply our stem() function to every movie's tags.
#
# This normalizes the words in all movies by converting
# different forms of words into their common stem/root form.
#
# Example:
# "loved loving loves"
#        ↓
# "love love love"
#
# This makes text similarity more meaningful for our
# movie recommendation system.

new_df['tags'] = new_df['tags'].apply(stem)
```

```python
# CountVectorizer converts our movie text (tags) into numerical vectors
# so that a machine learning algorithm can work with the textual data.

from sklearn.feature_extraction.text import CountVectorizer

# Create a CountVectorizer object.
# max_features=5000 means we keep at most 5,000 word features
# from all the words present in our movie tags.
#
# stop_words='english' removes common English words such as
# "the", "is", "a", "and", etc., because they usually provide
# little useful information for comparing movies.

cv = CountVectorizer(
    max_features=5000,
    stop_words='english'
)

# fit_transform() does two things:
#
# 1. FIT:
#    It analyzes all movie tags and learns the vocabulary/features
#    that will be used for vectorization.
#
# 2. TRANSFORM:
#    It converts every movie's text into a numerical vector
#    based on that learned vocabulary.
#
# Each column represents one word/feature.
# Each row represents one movie.
#
# If a movie contains a particular word, its value will be > 0.
# If the word does not appear in that movie, its value will be 0.
#
# Example:
# Vocabulary = [action, comedy, hero, romance]
#
# Movie 1: "action hero"
# → [1, 0, 1, 0]
#
# Movie 2: "comedy romance"
# → [0, 1, 0, 1]
#
# Since CountVectorizer COUNTS words, repeated words can have values
# greater than 1 (e.g., [2, 0, 1, 0]).
#
# .toarray() converts the resulting sparse matrix into a NumPy array.

vectors = cv.fit_transform(new_df['tags']).toarray()
```

```python
from sklearn.metrics.pairwise import cosine_similarity

# We have converted every movie into a vector based on its important words/features.
# For example, if our features are [action, romance, comedy]:
#
# Movie A → [1, 0, 1]
# Movie B → [1, 1, 1]
#
# 1 means that the feature/word is present in the movie,
# and 0 means that it is not present.
#
# Now we need to compare these movie vectors to find which movies are similar.
# cosine_similarity() compares two vectors by looking at the angle between them.
# Smaller angle → more similar movies
# Larger angle → less similar movies
```

```python
similarity = cosine_similarity(vectors)

# 'vectors' contains the vector representation of all our movies.
#
# cosine_similarity(vectors) compares EVERY movie with EVERY other movie.
#
# So if we have 4,806 movies, we get a 4,806 × 4,806 similarity matrix.
#
# Each row represents one movie.
# Each column represents another movie.
#
# similarity[i][j] tells us how similar movie i is to movie j.
#
# A value close to 1 → movies are very similar.
# A value close to 0 → movies have very little similarity.
#
# The diagonal values are 1 because every movie is 100% similar to itself.
```

```python
# MOVIE
#   ↓
# Important words/features
#   ↓
# Convert into numbers
#   ↓
# Vector
#   ↓
# Compare vector with other movie vectors
#   ↓
# Cosine Similarity
#   ↓
# Similarity Score
#   ↓
# Find movies with highest scores
#   ↓
# Recommend those movies
```

```python
# Get the similarity scores for movie at index 1
# compared with ALL 4,806 movies.
#
# The first value tells how similar movie 1 is to movie 0.
# The second value tells how similar movie 1 is to itself.
# The third value tells how similar movie 1 is to movie 2.
# And so on...
#
# The value 1 appears for the movie itself because
# a movie is completely similar to itself.
#
# These similarity scores will later help us find the most similar movies
# and recommend them to the user.
similarity[1]
```

```python
def recommend(movie):
    movie_index = new_df[new_df['title'] == movie].index[0]
    distances = similarity[movie_index]
    movies_list = sorted(list(enumerate(distances)), reverse=True, key=lambda x:x[1])[1:6]

    for i in movies_list:
        print(new_df.iloc[i[0]].title)
```
